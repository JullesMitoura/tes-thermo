```python
from tes_thermo.thermo_agent import ThermoAgent
from tes_thermo.utils import VectorSearch
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
load_dotenv(override=True)

# Create OpenAI client
openai_client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# Model names
llm_model_name = "gpt-4o-mini"  # or your preferred model
embedding_model_name = "text-embedding-3-small"  # or your preferred embedding model
```

```python
# Create vector search (optional, for RAG)
vsearch = VectorSearch.from_documents(
    document_paths=['example.pdf'],  # replace with your document paths (pdf only...)
    openai_client=openai_client,
    model_name=embedding_model_name
)

# Create ThermoAgent
agent = ThermoAgent(
    openai_client=openai_client,
    llm_model_name=llm_model_name,
    embedding_model_name=embedding_model_name,
    vsearch=vsearch
)
```

```python
prompt = """
Simulate the methane steam reforming process considering CH4, CO, CO2, H2, and H2O.
The process should be simulated at 1200 K and 1 bar.
Assume a feed consisting of 0.5 mole of CH4 and 1 moles of H2O.
Present the results in a table and discuss how I can improve the dynamic formation process and the solutions involved in the process.
"""
res = agent.chat(prompt=prompt)
print(res['output'])
```
